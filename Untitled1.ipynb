```python
import pandas as pd
import numpy as np
np.random.seed(42)

months = pd.date_range(start='2025-01-01', periods=12, freq='MS')
channels = ['social', 'search', 'email', 'paid_ad']
data = []

for month in months:
    seasonality = 1.0 if month.month <=9 else 1.2  # Q4 boost
    for channel in channels:
        if channel == 'social':
            spend = np.random.normal(20000, 2000)  # $20K avg
            cpa = 50
            avg_mrr = 49
            retention = 0.80
            lifetime = 8
        elif channel == 'search':
            spend = np.random.normal(15000, 1500)
            cpa = 40
            avg_mrr = 52
            retention = 0.85
            lifetime = 10
        elif channel == 'email':
            spend = np.random.normal(10000, 1000)
            cpa = 30
            avg_mrr = 49.55
            retention = 0.82
            lifetime = 9
        else:  # paid_ad
            spend = np.random.normal(15000, 1500)
            cpa = 100
            avg_mrr = 49.16
            retention = 0.78
            lifetime = 7
        
        users_acquired = (spend / cpa) * seasonality
        total_mrr_contribution = users_acquired * avg_mrr
        ltv = avg_mrr * lifetime * retention
        net_lift_per_1k = (total_mrr_contribution - spend) / (spend / 1000)
        
        data.append({
            'month': month,
            'channel': channel,
            'spend': spend,
            'users_acquired': users_acquired,
            'avg_mrr_per_user': avg_mrr,
            'total_mrr_contribution': total_mrr_contribution,
            'cpa': cpa,
            'retention_rate': retention,
            'lifetime_months': lifetime,
            'ltv': ltv,
            'seasonality_factor': seasonality,
            'net_mrr_lift_per_1k': net_lift_per_1k
        })

df = pd.DataFrame(data)
df.to_csv('channel_mrr_projections.csv', index=False)
print("Dataset created: channel_mrr_projections.csv")
print(df.head(8))  # First 2 months for all channels
```
