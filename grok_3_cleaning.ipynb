```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)

base_date = datetime(2024, 1, 1)
devices = ["Desktop", "Mobile", "Tablet"]
referrers = ["Google", "Direct", "Social", "Email", "Referral"]

data = []
for i in range(1000):
    session_id = f"SESS{i+1}"
    user_id = f"USER{random.randint(1, 200)}" if random.random() > 0.05 else ""
    timestamp = base_date + timedelta(seconds=random.randint(0, 30*24*60*60))  # 30 days
    date_format = random.choice(["%Y-%m-%d %H:%M:%S", "%m/%d/%Y %H:%M:%S"])
    timestamp = timestamp.strftime(date_format) if random.random() > 0.02 else "2024-13-01 10:00:00"
    page_url = random.choice(["home", "pricing", "blog", "signup"]) if random.random() > 0.1 else "http:/page"
    referrer = random.choice(referrers) if random.random() > 0.1 else ""
    device = random.choice(devices)
    time_on_page = random.uniform(-10, 300) if random.random() > 0.05 else random.choice(["ten", "zero"])
    bounce = random.choice(["Yes", "No", "Y", "N"]) if random.random() > 0.08 else ""

    data.append([session_id, user_id, timestamp, page_url, referrer, device, time_on_page, bounce])

df = pd.DataFrame(data, columns=["SessionID", "UserID", "Timestamp", "PageURL", 
                                 "Referrer", "DeviceType", "TimeOnPage", "Bounce"])
df.to_csv("website_analytics.csv", index=False)
print("Saved as 'website_analytics.csv'")
```

```python
df_clean['Timestamp'] = df_clean['Timestamp'].where(df_clean['Timestamp'].notna(), df_clean['TimestampOrginal'])
df_clean
```

```python
import re

def fix_us_dates(timestamp):
    match = re.match(r'(\d{4})-(\d{2})-(\d{2})\s(\d{2}:\d{2}:\d{2})', timestamp)
    if match:
        year, month, day, time = match.groups()
        if int(month) > 12:
            return f'{year}-{day}-{month} {time}'
        return timestamp

df_clean['Timestamp'] = df_clean['Timestamp'].apply(fix_us_dates)
df_clean['Timestamp'] = pd.to_datetime(df_clean['Timestamp'], errors='coerce')
df_clean
```

```python
df_clean = df_clean[~df_clean['PageURL'].str.contains('http:/page', case=False, na=False)].reset_index(drop=True)
df_clean
```
